```python
# ============================================
# SIMULADOR DE INTELIGENCIA ARTIFICIAL
# SOBRE CRIPTOMONEDAS
# Nivel: Principiante
# ============================================

print("===================================")
print("   IA DE CRIPTOMONEDAS")
print("===================================")

usuario = input("Ingresa tu nombre: ")

while True:

    print("\n===== MENÚ PRINCIPAL =====")
    print("1. Analizar Bitcoin")
    print("2. Analizar Ethereum")
    print("3. Analizar Dogecoin")
    print("4. Recomendación de inversión")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")

    # --------------------------------
    # BITCOIN
    # --------------------------------
    if opcion == "1":

        print("\n--- ANÁLISIS DE BITCOIN ---")

        precio_str = input("Ingresa el precio actual de Bitcoin: ")
        precio = float(precio_str.replace(',', '')) # Fix: Remove comma before converting to float
        tendencia = input("¿La tendencia es alta, media o baja?: ")

        if precio >= 70000 and tendencia == "alta":
            print("\nIA: Bitcoin tiene una tendencia MUY FUERTE.")
            print("IA: Posible oportunidad de compra.")

        elif precio >= 50000 and tendencia == "media":
            print("\nIA: Bitcoin está estable.")
            print("IA: Se recomienda observar el mercado.")

        else:
            print("\nIA: Bitcoin presenta riesgo.")
            print("IA: Se recomienda precaución.")

    # --------------------------------
    # ETHEREUM
    # --------------------------------
    elif opcion == "2":

        print("\n--- ANÁLISIS DE ETHEREUM ---")

        precio_str = input("Ingresa el precio actual de Ethereum: ")
        precio = float(precio_str.replace(',', '')) # Fix: Remove comma before converting to float
        volumen = input("¿El volumen es alto o bajo?: ")

        if precio >= 4000 and volumen == "alto":
            print("\nIA: Ethereum tiene buen rendimiento.")
            print("IA: Puede subir más.")

        elif precio >= 2000 and volumen == "alto":
            print("\nIA: Ethereum está creciendo lentamente.")

        else:
            print("\nIA: Ethereum tiene poca actividad.")
            print("IA: Mercado inestable.")

    # --------------------------------
    # DOGECOIN
    # --------------------------------
    elif opcion == "3":

        print("\n--- ANÁLISIS DE DOGECOIN ---")

        noticias = input("¿Hay noticias positivas? (si/no): ")
        comunidad = input("¿La comunidad está activa? (si/no): ")

        if noticias == "si" and comunidad == "si":
            print("\nIA: Dogecoin podría aumentar de valor.")

        elif noticias == "si" and comunidad == "no":
            print("\nIA: Hay interés, pero poca actividad.")

        else:
            print("\nIA: Dogecoin tiene baja probabilidad de crecimiento.")

    # --------------------------------
    # RECOMENDACIÓN
    # --------------------------------
    elif opcion == "4":

        print("\n--- RECOMENDACIÓN IA ---")

        dinero_str = input("¿Cuánto dinero deseas invertir?: ")
        dinero = float(dinero_str.replace(',', '')) # Fix: Remove comma before converting to float
        riesgo = input("¿Tu riesgo es bajo, medio o alto?: ")

        if dinero >= 10000 and riesgo == "alto":
            print("\nIA: Puedes invertir en criptomonedas volátiles.")
            print("IA: Recomendación -> Bitcoin y Dogecoin.")

        elif dinero >= 5000 and riesgo == "medio":
            print("\nIA: Inversión moderada.")
            print("IA: Recomendación -> Ethereum.")

        elif dinero < 5000 and riesgo == "bajo":
            print("\nIA: Invierte poco y con cuidado.")
            print("IA: Observa el mercado antes de invertir.")

        else:
            print("\nIA: Perfil de inversión no identificado.")

    # --------------------------------
    # SALIR
    # --------------------------------
    elif opcion == "5":

        print("\nGracias por usar la IA de criptomonedas.")
        print("Hasta luego", usuario)
        break

    # --------------------------------
    # OPCIÓN INVÁLIDA
    # --------------------------------
    else:
        print("\nOpción incorrecta.")
        print("Intenta nuevamente.")
```
